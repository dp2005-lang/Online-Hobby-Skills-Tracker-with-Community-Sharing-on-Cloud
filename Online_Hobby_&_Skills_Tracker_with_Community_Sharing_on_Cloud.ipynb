```python
# ============================================================
# HOBBY & SKILLS TRACKER — ONE-CELL GOOGLE COLAB VERSION
# Paste/run this ONE cell in Google Colab.
# ============================================================

import base64
import subprocess
import time
from pathlib import Path
from google.colab import output

# 1. Recreate the complete HTML/CSS/JavaScript application.
HTML_DATA = """"""

app_file = Path("/content/hobby_tracker.html")
app_file.write_bytes(base64.b64decode(HTML_DATA))

# 2. Start a local web server from the Colab runtime.
PORT = 8000

# Kill an older instance if the cell was previously run.
subprocess.run(
    ["bash", "-lc", f"fuser -k {PORT}/tcp >/dev/null 2>&1 || true"],
    check=False,
)

server = subprocess.Popen(
    ["python3", "-m", "http.server", str(PORT), "--directory", "/content"],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL,
)

time.sleep(1)

# 3. Display the complete application directly inside Colab.
print("✅ Hobby & Skills Tracker is running.")
print("Demo login: alice / demo1234")

output.serve_kernel_port_as_iframe(
    PORT,
    path="/hobby_tracker.html",
    width="100%",
    height=900,
)
```
